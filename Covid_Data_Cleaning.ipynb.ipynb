```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Environment ready")
```

```python
import pandas as pd
file_path = r"C:\Users\sandr\OneDrive\Covid Analysis\Covid Data.csv\Covid Data.csv"
cd = pd.read_csv(file_path)
```

```python
cd.head()
```

```python
cd.info()
```

```python
cd.shape
```

```python
cd["DATE_DIED"]=pd.to_datetime(cd["DATE_DIED"])
```

```python
cd["DATE_DIED"]=pd.to_datetime(cd["DATE_DIED"], format="%d/%m/%y")
```

```python
cd["DATE_DIED"]=pd.to_datetime(cd["DATE_DIED"], dayfirst = True, errors="coerce")
```

```python
cd.info()
```

```python
cd["DATE_DIED"].head()
```

```python
cd.isna().sum()
```

```python
cd["SEX"] = cd["SEX"].replace({1:"Female",2:"Male"})
```

```python
cd.head()
```

```python
cd["PATIENT_TYPE"] = cd["PATIENT_TYPE"].replace({1:"returned home",2:"hospitalization"})
```

```python
cd.head()
```

```python
cd["covid_status"] = cd["CLASIFFICATION_FINAL"].apply(
    lambda x: "Covid Positive" if x in [1,2,3] else "Covid negative / Inconclusive" )
```

```python
cd.head()
```

```python
cols_to_clean = cd.select_dtypes(include=["int64"]).columns
```

```python
exclude_cols = [
    "USMER"
    "MEDICAL_UNIT"
    "AGE"
    "CLASIFFICATION_FINAL"
]
```

```python
cols_to_clean = [ col for col in cd.columns
                  if col not in exclude_cols
                ]
```

```python
import numpy as np
cd[cols_to_clean]=cd[cols_to_clean].replace({97:np.nan,98: np.nan, 99: np.nan})
```

```python
cd[cols_to_clean]=cd[cols_to_clean].replace({1:"yes",2:"no"})
```

```python
cd["INTUBED"].value_counts(dropna=False)
```

```python
cd.head()
```

```python
cd_clean = cd.copy()
```

```python
cd_clean.shape
cd_clean.describe(include ="all")
```

```python
cd_clean["covid_status"].value_counts().plot(kind="bar")
plt.title("covid test results distribution")
plt.xlabel("covid status")
plt.ylabel("number of patients")
plt.show()
```

```python
cd_clean["SEX"].value_counts().plot(kind="pie")
plt.title("covid status by gender")
plt.xlabel("sex")
plt.ylabel("count")
plt.show()
```

```python
cd["AGE"] = pd.to_numeric(cd["AGE"],errors="coerce")

cd["Age_group"] = pd.cut(
    cd["AGE"] , 
    bins =[0,18,30,45,60,75,120],
    labels=["0-18", "19-30", "31-45", "46-60", "61-75", "75+"]
)
```

```python
cd["Age_group"].value_counts()
```

```python
ax = cd["Age_group"].value_counts().sort_index().plot(kind="bar")

for p in ax.patches:
    ax.annotate(
        str(p.get_height()),
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha="center",
        va="bottom"
    )
    plt.title("Patients by age group")
    plt.xlabel("Age_group")
    plt.ylabel("count")
```

```python
pd.crosstab(cd_clean["DIABETES"], cd_clean["covid_status"]).plot(kind="bar")
plt.title("Covid Status by Diabetes")
plt.xlabel("Diabetes")
plt.ylabel("Count")
plt.show()
```

```python
cd.to_csv("covid_cleaned_data.csv",index=False)
```
